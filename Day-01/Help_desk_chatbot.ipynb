```python
SYSTEM = """
You are a helpful Campus Helpdesk AI Assistant.

Facts:
- Our college provides undergraduate courses.
- Students can contact the college office for admission details.
- Students should contact library staff for library queries.
- Students can contact the administration for ID card issues.

Rules:
- Answer politely.
- Answer only using the facts provided.
- Never invent college information.
- If you don't know something, say:
  "Sorry, I don't have that information.
   Please contact the college office."

Format:
- Keep answers simple.
- Reply in under 60 words.
"""
```

```python
def ask(question):

    response = llm.invoke([
        ("system", SYSTEM),
        ("human", question)
    ])

    return response.content
```

```python
import os
from getpass import getpass

os.environ["GROQ_API_KEY"] = getpass("Enter your new Groq API Key: ")
```

```python
print("API Key Loaded:", bool(os.environ.get("GROQ_API_KEY")))
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    api_key=os.environ["GROQ_API_KEY"],
    temperature=0
)

print("Groq connection initialized!")
```
